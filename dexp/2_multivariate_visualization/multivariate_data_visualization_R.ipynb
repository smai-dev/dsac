```r
# Save the plot to a PNG file with increased width
png("ggparcoord_parallel_plot.png", width = 1200, height = 800)  # Adjust width and height as needed
ggparcoord(data = iris, 
           columns = 1:4, 
           groupColumn = 5, 
           scale = "std", 
           showPoints = TRUE)
dev.off()  # Close the PNG device
```

```r
# Adjust the plot window size using par()
par(mar = c(5, 5, 4, 2) + 0.1)  # Adjusts margins, this can help with the width of the plot
par(cex = 1.2)  # Adjusts the text size if needed

# Plot the parallel coordinates plot
parcoord(iris[, 1:4], col = as.factor(iris$Species), 
         main = "Parallel Coordinates Plot of Iris Dataset", 
         lwd = 2, 
         pch = 19)
```

```r
png("parcoord_parallel_plot.png", width = 1200, height = 800)
parcoord(iris[, 1:4], col = as.factor(iris$Species), 
         main = "Parallel Coordinates Plot of Iris Dataset", 
         lwd = 2, 
         pch = 19)
dev.off()  # Close the PNG device
```

```r
# Scale the iris data
iris_scaled <- scale(iris[,1:4])

# Create a color mapping for the species annotation
annotation_colors <- list(Species = c(setosa = "red", versicolor = "green", virginica = "blue"))

# Generate the heatmap
pheatmap(iris_scaled, 
         cluster_rows = TRUE, 
         cluster_cols = TRUE, 
         annotation_col = data.frame(Species = iris$Species),
         annotation_colors = annotation_colors)
```
