# 1. Properties of a Dataset

A dataset and its attributes can be described in terms of **Dimensionality**, **Density**, and **Resolution**.

- **Dimensionality:**  
  The number of attributes each data object possesses.  
  - *Insight:* A very low dimensionality may oversimplify the data, while high dimensionality can hinder pattern detection due to the "curse of dimensionality."

- **Density:**  
  The proportion of non-missing values in the dataset.  
  - *Insight:* Sparse datasets (low density) can limit the choice of mining algorithms.

- **Resolution:**  
  The scale or granularity at which an attribute is recorded. For example, height might be measured in centimeters or meters.  
  - *Insight:* Choosing the appropriate resolution is crucial since it impacts the variability and noise in the data. When merging datasets, ensure that attributes share the same resolution.

# 3. Types of Datasets

Different data types require different handling and analysis methods:

1. **Record Data:**  
   Data that consists of a collection of records with a fixed set of attributes.
   
2. **Transaction Data:**  
   A special type of record data where each record (transaction) is a set of items.  
   *Example:* Grocery store purchases where each transaction consists of multiple items (each with a Transaction ID or TID).

3. **Data Matrix:**  
   Represented as an \( m \times n \) matrix (where \( m \) is the number of objects and \( n \) is the number of attributes), allowing data objects to be viewed as points in a multi-dimensional space.

4. **Document Data:**  
   Each document is represented as a term vector. The value in each dimension corresponds to the frequency of a term, typically stored in a sparse matrix format.

5. **Graph Data:**  
   Used for data with relationships such as HTML links or network connections.

6. **Chemical Data:**  
   For example, molecular structures (e.g., Benzene, \( \text{C}_6\text{H}_6 \)).

7. **Ordered Data:**  
   Includes data with inherent ordering:  
   - **Sequential/Temporal Data:** Each record has a timestamp.  
   - **Time Series Data:** A series of measurements taken at regular or irregular intervals, e.g., the Dow Jones Industrial Average or GDP.  
   - **Spatial Data:** Data with geographical attributes.  
   - **Sequence Data:** Where the order is significant but not tied to time, e.g., genetic sequences.

# 4. Attribute Values and Types

## 4.1. Attribute Values

- **Attribute Values:**  
  Numbers or characters assigned to an attribute. The same attribute can have different value types (e.g., height measured in feet or meters).

## 4.2. Types of Attributes

1. **Nominal:**  
   - *Examples:* ID numbers, eye color, zip codes.  
   - *Operations:* Mode, entropy, contingency correlation.
   
2. **Ordinal:**  
   - *Examples:* Rankings (e.g., taste ratings from 1 to 10).  
   - *Operations:* Median, percentiles, rank correlation.  
   - *Transformation:* Must preserve the order (monotonic function).

3. **Interval:**  
   - *Examples:* Calendar dates, temperatures in Celsius.  
   - *Operations:* Mean, standard deviation, Pearson's correlation, t and F tests.  
   - *Transformation:* Linear transformation:  
     $$
     \text{new\_value} = a \times \text{old\_value} + b
     $$
     where \(a\) and \(b\) are constants.

4. **Ratio:**  
   - *Examples:* Temperature in Kelvin, length, time, counts.  
   - *Operations:* Same as interval plus geometric mean, harmonic mean, percent variation.  
   - *Transformation:* Multiplicative:  
     $$
     \text{new\_value} = a \times \text{old\_value}
     $$

### 4.3. Discrete vs. Continuous Attributes

- **Discrete Attributes:**  
  Finite or countably infinite set of values (e.g., zip codes, counts).  
  *Note:* Binary attributes are a subset of discrete attributes.

- **Continuous Attributes:**  
  Real-numbered values, typically approximated with finite precision (e.g., temperature, height).

### 4.4. Scalar vs. Vector

- **Scalar:**  
  A single value per attribute (e.g., age).
- **Vector:**  
  Multiple values per attribute (e.g., geographical coordinates).

# 5. Data Types: Operations & Transformations

An attribute's type dictates the valid operations and transformations:

- **Nominal & Categorical:**  
  - *Operations:* Mode, entropy, contingency correlation.  
  - *Transformations:* One-to-one mappings that do not change the information content.
  
- **Ordinal:**  
  - *Operations:* Median, percentiles, rank correlation.  
  - *Transformations:* Must preserve the ordering.

- **Interval:**  
  - *Operations:* Mean, standard deviation, Pearson's correlation.  
  - *Transformations:* Linear transformations \( \text{new\_value} = a \times \text{old\_value} + b \).

- **Ratio:**  
  - *Operations:* Same as interval plus geometric and harmonic means, percent variation, t tests.  
  - *Transformations:* Multiplicative scaling \( \text{new\_value} = a \times \text{old\_value} \).

# 6. Data Quality

Data quality issues can greatly affect the results of your analysis. Common problems include:

- **Noise:**  
  Random variations or errors in data.  
  - *Detection:* Increased variance or visible outliers in histograms.
  
- **Outliers:**  
  Values that significantly differ from other observations.  
  - *Insight:* Outliers may indicate errors or valuable anomalies.

- **Missing Values:**  
  - *Reasons:* Data not collected, inapplicable attributes, or entry errors.
  - *Handling Techniques:*  
    - Eliminate records with missing values.
    - Impute missing values using statistical estimates.
    - Ignore during analysis or replace with weighted probabilities.

- **Duplicate Data:**  
  Redundant records, often emerging during the merging of heterogeneous data sources.

- **Inconsistent Data:**  
  Typographical errors or conflicting entries that may lead to duplicates or misinterpretations.

- **Bias:**  
  Systematic error where the sample is not representative of the population.  
  - *Example:* A dataset on loan defaults that only includes borrowers from a particular region.

### Precision, Bias, and Accuracy

- **Precision:**  
  Consistency of repeated measurements, typically measured by standard deviation.
  
- **Bias:**  
  Systematic deviation, measured by the arithmetic mean.
  
- **Accuracy:**  
  Overall degree of error in measurement.

# 7. Data Preprocessing Techniques

Preprocessing is essential to ensure data quality and to prepare data for mining. Techniques include:

## 7.1. Aggregation

- **Purpose:**  
  Reduce data size, smooth noise, and achieve a higher level of abstraction.
- **Example:**  
  Aggregating cities into regions or states.

## 7.2. Sampling

- **Purpose:**  
  Obtain a representative subset of the data when the full dataset is too large to process.
- **Types:**  
  - **Simple Random Sampling:** Each item has an equal chance.  
  - **Stratified Sampling:** Data is divided into partitions before sampling.
  
- **Insight:**  
  A representative sample closely mirrors the original data properties.

## 7.3. Dimensionality Reduction

- **Purpose:**  
  Mitigate the curse of dimensionality and reduce computational costs.
- **Techniques:**  
  - Principal Component Analysis (PCA)
  - Singular Value Decomposition (SVD)
  - Other supervised or non-linear methods

## 7.4. Feature Subset Selection

- **Purpose:**  
  Remove redundant and irrelevant features.
- **Techniques:**  
  - **Brute-Force:** Test all possible subsets.
  - **Embedded:** Feature selection built into the mining algorithm.
  - **Filter/Wrapper Approaches:** Select features based on statistical measures or algorithm performance.

## 7.5. Feature Creation

- **Purpose:**  
  Construct new features that better capture the underlying information.
- **Methods:**  
  - Feature Extraction
  - Mapping Data to a New Space
  - Feature Construction by combining existing features

## 7.6. Discretization and Binarization

- **Discretization:**  
  Transform continuous attributes into categorical ones.
- **Binarization:**  
  Convert attributes (discrete or continuous) into binary variables.

## 7.7. Variable Transformation and Normalization

- **Variable Transformation:**  
  Apply functions such as \( \log(x) \), \( \sqrt{x} \), \( e^x \), or \( \frac{1}{x} \) to adjust data distribution.
  
- **Normalization:**  
  Standardize the data so that it has a mean of 0 and a standard deviation of 1.  
  - **Formula:**  
    $$
    x' = \frac{x - \text{mean}}{\text{standard deviation}}
    $$
  - *Note:* Outliers can significantly affect the mean and standard deviation. Alternatives such as using the median or absolute deviation may be used.

- **Attribute Transformation:**  
  Map the entire set of attribute values to a new set while preserving key properties (e.g., order for ordinal attributes).

# 8. When to Use Different Techniques in Data Mining

1. **Understanding and Data Quality:**  
   - **Exploratory Data Analysis (EDA):**  
     Use descriptive statistics (mean, median, mode, variance, skewness, and kurtosis) to assess the overall quality and distribution of the data.
   - **Data Quality Checks:**  
     Detect noise, missing values, duplicates, and biases early to decide if cleaning or transformation is needed.

2. **Preprocessing and Transformation:**  
   - **Sampling:**  
     When the dataset is too large or processing all data is infeasible.
   - **Aggregation and Dimensionality Reduction:**  
     To reduce noise and the computational burden, especially when facing high-dimensional data.
   - **Feature Selection and Creation:**  
     Use when certain features are redundant or when new combinations of features might reveal hidden patterns.
   - **Normalization and Transformation:**  
     Essential when different attributes have varied scales or non-normal distributions. Transformations help meet assumptions required by many statistical algorithms.

3. **Data Type Specific Operations:**  
   - **Nominal & Categorical:**  
     Focus on mode, entropy, and contingency measures.
   - **Ordinal:**  
     Use median and rank-based measures while preserving the order in transformations.
   - **Interval and Ratio:**  
     Apply arithmetic operations (mean, standard deviation) and linear transformations. Ratio data allows for additional measures such as geometric and harmonic means.
