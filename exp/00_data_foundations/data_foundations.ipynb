# Data Foundations: Understanding and Preparing Structured Data

## Scope

Data exploration begins by identifying what each observation represents, how the fields were collected, and which transformations are meaningful for those fields. This chapter provides the conceptual background for the practical R and Python notebooks in this collection.

## 1. Describe the dataset before modeling

**Rows and columns.** Records usually represent entities, transactions, or measurements; columns encode their properties. Dimensionality is the number of fields, whereas sparsity refers to absent or zero-valued information. Resolution describes how finely a measurement is recorded and should not be confused with its measurement unit.

**Ask first:** What is the observational unit? Are the records independent? Are missing values actually missing, or do they mean “not applicable”? Are all timestamps and units consistent?

## 2. Recognize the data structure

- **Tabular records:** each observation has a set of named fields.
- **Transactions:** sets of items or events, often represented as sparse indicators.
- **Numerical matrices:** rows may be interpreted as points in a feature space.
- **Documents:** text represented through terms, vectors, or embeddings.
- **Graphs:** entities connected through explicit relations.
- **Sequences and time series:** ordering and temporal dependence influence valid operations.
- **Spatial and scientific data:** locations, continuous measurements, or domain-specific structures require contextual interpretation.

## 3. Match the operation to the attribute

| Attribute | Example | Suitable summaries | Caution |
|---|---|---|---|
| Nominal | Location category | Counts, mode, proportions | Numeric codes do not imply distance |
| Ordinal | Satisfaction level | Counts, median, percentiles | Preserve rank order |
| Interval | Temperature in °C | Differences, mean, standard deviation | Ratios are not intrinsically meaningful |
| Ratio | Duration, mass | Differences, mean, ratios | Requires a meaningful zero |

Integer-valued variables can be categorical; continuous measurements may be stored as rounded integers. The data type alone does not establish statistical meaning.

## 4. Inspect data quality systematically

Check completeness, duplicates, inconsistent labels, illegal values, inconsistent measurement units, and possible collection bias. Distinguish **precision** (repeatability), **accuracy** (proximity to the true value), and **bias** (systematic deviation). Unusual observations require investigation, not automatic removal.

## 5. Prepare data with a clear purpose

1. **Aggregation:** combine measurements at a coarser level when the modeling question permits it.
2. **Sampling:** choose a representative subset; stratification can preserve group proportions.
3. **Feature selection:** retain useful attributes while controlling redundancy and leakage.
4. **Feature creation:** construct interpretable quantities from existing measurements.
5. **Discretization and encoding:** convert fields for algorithms while respecting their meaning.
6. **Scaling:** apply standardization or normalization where scale-sensitive methods require it.
7. **Dimensionality reduction:** summarize correlated dimensions, recognizing possible interpretability loss.

When a model is evaluated, fit data-dependent transformations on training data and apply them to validation/test data. Fitting encoders or scalers to all observations beforehand can leak information.

## 6. Connect exploration with CRISP-DM

The data-understanding phase includes acquiring the data, describing its contents, inspecting quality, and exploring relationships relevant to the objective. Findings can send the project back to business understanding or forward to preparation, modeling, and evaluation.

## Practical checklist

- Define the population, sampling method, and observational unit.
- Inventory fields, ranges, missing entries, and category frequencies.
- Plot meaningful univariate and multivariate relationships.
- Record anomalies and unresolved assumptions.
- Explain any transformation before applying it.
- Preserve a reproducible path from source data to results.

**Outcome:** A defensible description of what the dataset can support—and which questions still require investigation.
